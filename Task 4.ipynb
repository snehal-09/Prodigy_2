```python
col_name=['ID','Entity','Sentiments','Context']
df=pd.read_csv('twitter_training.csv.zip',names=col_name)
```

```python
df.dropna(axis=0,inplace=True)
```

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df.shape
```

```python
sentiment_count=df['Sentiments'].value_counts()
sentiment_count
```

```python
plt.figure(figsize=(6,3))
sentiment_count.plot(kind='bar',color=['red','green','blue','yellow'])
plt.title('sentiment Distribution')
plt.xlabel('Number of Tweets')
plt.xticks(rotation=0)
plt.show()
```

```python
brand_data=df[df['Entity'].str.contains('Microsoft',case=False)]
brand_semantic_counts=brand_data['Sentiments'].value_counts()
brand_semantic_counts
```

```python
plt.figure(figsize=(6,6))
plt.pie(brand_semantic_counts,labels=brand_semantic_counts.index,autopct='%1.11f%%',startangle=140)
plt.show()
```
